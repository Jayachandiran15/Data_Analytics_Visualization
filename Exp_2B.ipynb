```python
import pandas as pd

# ============================
# Load the dataset
# ============================
df = pd.read_csv("data.csv")

# ============================
# Display dataset
# ============================
print("First 5 Rows:")
print(df.head())

print("\nLast 5 Rows:")
print(df.tail())

# ============================
# Dataset information
# ============================
print("\nDataset Information:")
df.info()

print("\nSummary Statistics:")
print(df.describe(include='all'))

# ============================
# Handle missing values
# ============================
numeric_columns = df.select_dtypes(include='number').columns
df[numeric_columns] = df[numeric_columns].fillna(df[numeric_columns].mean())

# ============================
# Create a new column
# ============================
if 'existing_column' in df.columns:
    df['new_column'] = df['existing_column'] * 2
    print("\nNew column created successfully.")

# ============================
# Series operations
# ============================
if 'existing_column' in df.columns:
    series = df['existing_column']
    print("\nSeries after adding 10:")
    print(series + 10)

# ============================
# Filter rows
# ============================
if {'existing_column', 'another_column'}.issubset(df.columns):
    filtered_df = df[
        (df['existing_column'] > 50) &
        (df['another_column'] < 100)
    ]

    print("\nFiltered DataFrame:")
    print(filtered_df)

# ============================
# Grouping and aggregation
# ============================
if {'category_column', 'numeric_column'}.issubset(df.columns):
    grouped = df.groupby('category_column')['numeric_column'].mean()

    print("\nMean by Category:")
    print(grouped)

# ============================
# Sorting
# ============================
if 'numeric_column' in df.columns:
    sorted_df = df.sort_values(by='numeric_column', ascending=False)

    print("\nSorted DataFrame:")
    print(sorted_df)

# ============================
# Boolean masking
# ============================
if 'numeric_column' in df.columns:
    masked_df = df[
        df['numeric_column'] > df['numeric_column'].median()
    ]

    print("\nRows Above Median:")
    print(masked_df)

# ============================
# Remove duplicates and missing values
# ============================
df.drop_duplicates(inplace=True)
df.dropna(inplace=True)

# ============================
# Create subset of columns
# ============================
if {'column1', 'column2'}.issubset(df.columns):
    subset_df = df[['column1', 'column2']]
    subset_df.to_csv("filtered_data.csv", index=False)

    print("\nSubset saved as 'filtered_data.csv'.")

# ============================
# Statistical calculations
# ============================
if 'numeric_column' in df.columns:
    print("\nStatistics for numeric_column")
    print("Total Sum           :", df['numeric_column'].sum())
    print("Mean                :", df['numeric_column'].mean())
    print("Standard Deviation  :", df['numeric_column'].std())
```
